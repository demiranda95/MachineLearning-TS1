## 2. Punto 1: Frameworks para analítica de datos

### 2.1 Contexto

El desarrollo de proyectos de analítica de datos y machine learning requiere un enfoque estructurado que permita abordar de manera sistemática las distintas etapas del proceso, desde la comprensión del problema hasta la evaluación de los resultados.

En este contexto, los frameworks de analítica de datos proporcionan una guía metodológica que facilita la organización del trabajo, la trazabilidad de las decisiones y la reproducibilidad del análisis. Estos marcos permiten reducir la improvisación, mejorar la calidad de los resultados y asegurar que el desarrollo esté alineado con los objetivos del negocio.

A continuación, se presentan tres frameworks ampliamente utilizados en proyectos de ciencia de datos: CRISP-DM, KDD y TDSP.

### 2.2 CRISP-DM (Cross Industry Standard Process for Data Mining)

CRISP-DM es uno de los frameworks más utilizados en proyectos de analítica de datos, debido a su flexibilidad y enfoque iterativo. Este modelo propone un ciclo compuesto por seis etapas principales:

1. **Comprensión del negocio**: definición del problema, objetivos y criterios de éxito.
2. **Comprensión de los datos**: exploración inicial de los datos y detección de problemas.
3. **Preparación de los datos**: limpieza, transformación e integración de las fuentes.
4. **Modelado**: selección de algoritmos y entrenamiento de modelos.
5. **Evaluación**: validación del modelo en función de los objetivos definidos.
6. **Despliegue**: implementación del modelo en un entorno operativo.

#### Ventajas
- Enfoque claro en la comprensión del problema de negocio.
- Naturaleza iterativa que permite ajustar el proceso.
- Amplia adopción en la industria.

#### Limitaciones
- No define herramientas específicas.
- Puede ser interpretado de forma general si no se aplica rigurosamente.

### 2.3 KDD (Knowledge Discovery in Databases)

El proceso KDD se centra en la extracción de conocimiento a partir de grandes volúmenes de datos. Sus etapas principales son:

1. Selección de datos
2. Preprocesamiento
3. Transformación
4. Data mining (modelado)
5. Interpretación y evaluación

#### Ventajas
- Fuerte énfasis en la preparación y transformación de datos.
- Orientado a la generación de conocimiento a partir de datos.

#### Limitaciones
- Menor enfoque en la comprensión del negocio.
- No contempla explícitamente el despliegue de modelos.

### 2.4 TDSP (Team Data Science Process)

TDSP es un framework desarrollado por Microsoft, orientado a equipos de trabajo en proyectos de ciencia de datos. Sus etapas principales incluyen:

1. Comprensión del negocio
2. Adquisición y comprensión de datos
3. Modelado
4. Despliegue
5. Comunicación de resultados

#### Ventajas
- Enfoque colaborativo.
- Integra buenas prácticas de ingeniería de software.
- Fuerte énfasis en la documentación y reproducibilidad.

#### Limitaciones
- Puede ser más rígido que otros enfoques.
- Requiere mayor formalización en proyectos pequeños.

### 2.5 Comparación de frameworks

| Criterio                     | CRISP-DM | KDD | TDSP |
|----------------------------|----------|-----|------|
| Enfoque en negocio         | Alto     | Bajo | Alto |
| Preparación de datos       | Alta     | Muy alta | Alta |
| Iteratividad               | Sí       | Parcial | Sí |
| Despliegue                 | Sí       | No | Sí |
| Trabajo colaborativo       | Medio    | Bajo | Alto |

### 2.6 Framework adoptado

Para el desarrollo de este trabajo se adopta el framework **CRISP-DM**, debido a su flexibilidad y a su fuerte alineación con problemas reales de negocio.

En particular, este enfoque permite estructurar el análisis en etapas claras que se reflejan directamente en la organización del notebook:

- Comprensión del problema (Introducción e hipótesis)
- Comprensión y exploración de datos (EDA)

El carácter iterativo de CRISP-DM resulta especialmente relevante en este caso, ya que permite ajustar decisiones como el tratamiento de variables complejas (por ejemplo, MarkDown) en función de la evidencia obtenida durante el análisis.

### 3.2 Carga y comprensión de fuentes

El análisis se basa en tres conjuntos de datos principales:

- **train.csv**: contiene las ventas semanales por tienda y departamento.
- **features.csv**: incluye variables externas como temperatura, precio del combustible, indicadores económicos y variables promocionales (MarkDown).
- **stores.csv**: contiene características de las tiendas, como tipo y tamaño.

Cada dataset aporta información complementaria, por lo que es necesario integrarlos para construir una base analítica unificada.

```python
import pandas as pd

# Carga de datasets
train = pd.read_csv('/content/train.csv')
features = pd.read_csv('/content/features.csv')
stores = pd.read_csv('/content/stores.csv')

# Mostrar dimensiones
print("Train:", train.shape)
print("Features:", features.shape)
print("Stores:", stores.shape)
```

#### Observaciones iniciales

Se observa que los datos se encuentran separados en múltiples fuentes, lo que implica la necesidad de realizar un proceso de integración.

#### Análisis de estructura

A partir de la exploración inicial, se identifican las siguientes características:

- La variable objetivo es `Weekly_Sales`, de tipo numérico.
- La variable `Date` se encuentra en formato string, por lo que deberá convertirse a tipo fecha.

Estas observaciones anticipan la necesidad de realizar transformaciones y limpieza antes del modelado.

```python
# Conversión de fechas
train['Date'] = pd.to_datetime(train['Date'])
features['Date'] = pd.to_datetime(features['Date'])
```

### 3.2 Integración de datos

Para realizar el análisis, es necesario integrar las distintas fuentes de datos en una única base analítica.

El dataset **train** contiene la variable objetivo (`Weekly_Sales`) y será utilizado como base principal. A este se le incorporan:

- Variables externas desde **features**, utilizando las claves `Store`, `Date` e `IsHoliday`.
- Características de las tiendas desde **stores**, utilizando la clave `Store`.

Esta integración permite enriquecer cada observación con información contextual relevante, como condiciones económicas, promociones y atributos de la tienda.

```python
# Merge con features
df = pd.merge(train, features, on=['Store', 'Date', 'IsHoliday'], how='left')

print("Dimensión después de merge con features:", df.shape)
```

```python
# Merge con stores
df = pd.merge(df, stores, on='Store', how='left')

print("Dimensión final:", df.shape)
```

#### Validación de la integración

Una vez realizada la integración, es necesario verificar que el proceso no haya generado inconsistencias, tales como:

- Pérdida de registros
- Duplicación de observaciones
- Generación de valores faltantes inesperados

Dado que se utilizó un **left join**, se espera conservar todas las observaciones del dataset original (`train`), incorporando información adicional cuando esté disponible.

```python
# Verificar duplicados
print("Duplicados:", df.duplicated().sum())

# Verificar missing por columna
missing = df.isnull().sum().sort_values(ascending=False)
missing.head(10)
```

#### Observaciones

Tras la integración, se observa la presencia de valores faltantes principalmente en las variables asociadas a promociones (MarkDown), lo cual es consistente con la naturaleza de este tipo de datos.

Esto sugiere que no todas las semanas o tiendas presentan actividades promocionales, o bien que estas no siempre son registradas.

### 3.4 Análisis de valores faltantes

El análisis de valores faltantes es una etapa fundamental en la preparación de datos, ya que la forma en que estos son tratados puede afectar significativamente el desempeño del modelo.

En este caso, es importante no asumir automáticamente que los valores faltantes representan errores, sino evaluar su significado en el contexto del negocio. En particular, las variables de promoción (MarkDown) requieren un análisis más detallado, ya que la ausencia de datos puede estar asociada a la inexistencia de promociones.

El objetivo de esta sección es identificar la magnitud de los valores faltantes, comprender su origen y definir estrategias de tratamiento adecuadas para cada tipo de variable.


```python
missing = df.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]

missing
```

```python
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_pct = missing_pct.sort_values(ascending=False)
missing_pct[missing_pct > 0]
```

#### Observaciones generales

Se observa que los valores faltantes se concentran principalmente en las variables de promoción (`MarkDown1` a `MarkDown5`).

Este patrón no parece aleatorio, sino estructural, lo que sugiere que los valores faltantes podrían estar asociados a la ausencia de promociones en determinadas semanas o tiendas.

Por otro lado, las variables económicas y estructurales presentan pocos o ningún valor faltante, lo que indica una mayor consistencia en su registro.

#### Análisis específico de MarkDown

```python
markdown_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown4','MarkDown5']

for col in markdown_cols:
    print(f"\n{col}")
    print("Missing:", df[col].isnull().sum())
    print("Ceros:", (df[col] == 0).sum())
```

#### Interpretación de variables MarkDown

Las variables de promoción presentan una alta proporción de valores faltantes y ceros.

Esto sugiere que:

- Los valores faltantes no necesariamente corresponden a errores de registro.
- Es probable que representen la ausencia de promociones en determinadas observaciones.
- La distribución de estas variables es altamente asimétrica, con pocos valores positivos y algunos valores extremos.

En este contexto, imputar valores faltantes utilizando métodos tradicionales (media o mediana) no resulta apropiado, ya que distorsionaría la interpretación de la variable.

#### Estrategia de tratamiento

En base al análisis realizado, se adoptan las siguientes decisiones:

1. **Variables MarkDown**
   - Los valores faltantes se interpretan como ausencia de promoción.
   - Se reemplazan por 0 para reflejar este comportamiento.

2. **Otras variables**
   - En caso de existir valores faltantes, se evaluará su tratamiento según el contexto (imputación o eliminación).

Esta decisión permite mantener la coherencia con el fenómeno de negocio y evitar introducir sesgos artificiales en el modelo.

```python
# Reemplazo de NaN por 0 en variables MarkDown
df[markdown_cols] = df[markdown_cols].fillna(0)
```

### 3.5 Análisis de valores atípicos (Outliers)

El análisis de valores atípicos es fundamental para comprender comportamientos extremos en los datos y evaluar si estos corresponden a errores o a fenómenos reales del negocio.

En el contexto de ventas, valores altos no necesariamente representan anomalías, sino que pueden estar asociados a eventos como promociones, feriados o temporadas específicas.

Por lo tanto, el objetivo de esta sección es identificar la presencia de outliers y determinar si deben ser tratados, transformados o mantenidos para el modelado.

#### 1. Visualización de outliers en ventas

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Weekly_Sales'])
plt.title('Boxplot de Weekly Sales')
plt.show()
```

#### Observaciones

Se observa la presencia de valores extremos en la distribución de ventas, lo cual es consistente con el análisis previo de la variable objetivo.

Estos valores no parecen ser errores, sino eventos reales de alta demanda, posiblemente asociados a promociones o períodos específicos del año.

Por lo tanto, eliminar estos valores podría implicar la pérdida de información relevante para el modelo.

#### 2. Identificación cuantitativa (IQR)

```python
Q1 = df['Weekly_Sales'].quantile(0.25)
Q3 = df['Weekly_Sales'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Weekly_Sales'] < lower) | (df['Weekly_Sales'] > upper)]

print("Cantidad de outliers:", len(outliers))
print("Porcentaje:", len(outliers) / len(df) * 100)
```

#### Observaciones

Se identifica un porcentaje significativo de valores considerados atípicos según el criterio estadístico del rango intercuartílico (IQR).

Sin embargo, este criterio no distingue entre errores y eventos reales del negocio, por lo que su uso debe ser complementado con una interpretación contextual.

#### 3. Relación con promociones

```python
df['Has_MarkDown'] = (df[markdown_cols].sum(axis=1) > 0).astype(int)

sns.boxplot(x='Has_MarkDown', y='Weekly_Sales', data=df)
plt.title('Ventas con y sin promociones')
plt.show()
```

#### Observaciones

Se observa que las ventas asociadas a promociones presentan mayor dispersión y valores más altos.

Esto sugiere que una parte importante de los valores extremos está explicada por eventos promocionales, lo que refuerza la idea de que estos outliers corresponden a comportamiento real del sistema.

#### 4. Estrategia de tratamiento

En base al análisis realizado, se adopta la siguiente decisión:

- No eliminar los valores atípicos de la variable `Weekly_Sales`.
- Mantener estos valores, ya que representan eventos reales relevantes para el problema de predicción.
- Considerar transformaciones (por ejemplo, logaritmo) en etapas posteriores si el modelo lo requiere.

Esta decisión permite preservar la variabilidad del sistema y evitar la pérdida de información crítica.

|---------------|----------|-------------|

| Binario        | Simple, robusto, interpretable | Pierde información de magnitud |
| Rangos         | Captura niveles de promoción | Problemas por concentración en cero |

#### 5. Decisión de modelado

En base al análisis realizado, se opta por utilizar una representación **binaria** de las variables MarkDown, donde:

- 1 indica presencia de promoción
- 0 indica ausencia de promoción

Esta decisión se justifica por:

- Mayor estabilidad frente a valores extremos
- Mejor interpretabilidad
- Evidencia empírica de relación con las ventas
- Naturaleza dispersa de las variables originales

No obstante, en etapas posteriores se evaluará la posibilidad de complementar este enfoque con información adicional, como la intensidad de las promociones.

## 4. Ventanas de desarrollo, validación y test

En proyectos de machine learning, la forma en que se dividen los datos en conjuntos de entrenamiento, validación y prueba es fundamental para garantizar una evaluación adecuada del modelo.

En problemas con componente temporal, como el presente caso de predicción de ventas, no es apropiado utilizar particiones aleatorias, ya que esto podría introducir fuga de información (data leakage), al mezclar datos del futuro en el proceso de entrenamiento.

Por lo tanto, es necesario utilizar un enfoque basado en el tiempo, que respete la secuencia temporal de los datos.

### 4.1 Métodos de partición

Existen distintos enfoques para dividir los datos:

#### 1. Partición aleatoria (train/test split)
Consiste en dividir los datos de forma aleatoria.

- Ventaja: simple de implementar
- Desventaja: no respeta la temporalidad

#### 2. Validación cruzada (cross-validation)
Divide los datos en múltiples subconjuntos para evaluar el modelo.

- Ventaja: evaluación robusta
- Desventaja: no adecuada para series temporales sin modificaciones

#### 3. Partición temporal (time-based split)
Divide los datos respetando el orden temporal.

- Ventaja: evita fuga de información
- Desventaja: menor cantidad de datos para entrenamiento

Dado el contexto del problema, este último enfoque es el más adecuado.

### 4.2 Elección del método

Se opta por utilizar una partición basada en el tiempo, ya que:

- El problema presenta una estructura temporal clara
- El objetivo es predecir valores futuros a partir de datos históricos
- Se busca evitar data leakage

Este enfoque permite simular un escenario real, donde el modelo se entrena con información pasada y se evalúa en datos futuros.

### 4.3 Aplicación al caso

```python
# Ordenar por fecha
df = df.sort_values(by='Date')
```

```python
# Definir fechas de corte
train_end = '2012-06-01'
val_end = '2012-10-01'

train_df = df[df['Date'] <= train_end]
val_df = df[(df['Date'] > train_end) & (df['Date'] <= val_end)]
test_df = df[df['Date'] > val_end]

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

#### Observaciones

Se han definido tres conjuntos de datos:

- **Train**: utilizado para entrenar el modelo
- **Validation**: utilizado para ajustar hiperparámetros
- **Test**: utilizado para evaluar el desempeño final

Esta división respeta la estructura temporal del problema y permite simular un escenario real de predicción.

### 4.4 Visualización de cortes

#### Observaciones

La visualización confirma que la división respeta la continuidad temporal de los datos.

Esto es fundamental para evitar que el modelo acceda a información futura durante el entrenamiento, lo que podría generar resultados artificialmente optimistas.

## 5. Modelado base y comparación de alternativas

En esta etapa se construyen modelos predictivos con el objetivo de estimar las ventas semanales (`Weekly_Sales`).

El enfoque adoptado consiste en:

- Definir un conjunto de variables explicativas en base al análisis previo
- Construir un modelo base como referencia
- Evaluar modelos alternativos
- Comparar su desempeño utilizando métricas apropiadas

El objetivo no es únicamente obtener el mejor resultado numérico, sino comprender el comportamiento de los modelos y su capacidad de generalización.

### 5.1 Selección de variables

En base al análisis exploratorio, se seleccionan las siguientes variables para el modelado:

- Variables económicas: `Temperature`, `Fuel_Price`, `CPI`, `Unemployment`
- Variables de tienda: `Store`, `Dept`, `Size`
- Variable categórica de tienda: `Type` (a codificar mediante variables dummy)
- Variable de feriado: `IsHoliday`
- Variables temporales derivadas: `Year`, `Month`, `Week`
- Variables de promoción transformadas en formato binario: `MarkDown1_bin` a `MarkDown5_bin`

Se excluyen las variables originales de `MarkDown1` a `MarkDown5` debido a su alta asimetría, dispersión y sensibilidad a valores extremos, priorizando su versión binaria por mayor estabilidad e interpretabilidad.

```python
# Crear variables temporales sobre copias explícitas para evitar SettingWithCopyWarning
train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

for dataset in [train_df, val_df, test_df]:
    dataset.loc[:, 'Year'] = dataset['Date'].dt.year
    dataset.loc[:, 'Month'] = dataset['Date'].dt.month
    dataset.loc[:, 'Week'] = dataset['Date'].dt.isocalendar().week.astype(int)
```

### 5.2 Preparación de datos para modelado

Dado que `Type` es una variable categórica, se transforma mediante codificación one-hot.

Posteriormente, se alinean las columnas de los conjuntos de entrenamiento, validación y prueba para asegurar que todos tengan la misma estructura antes del modelado.

```python
# Codificación categórica de Type
train_df = pd.get_dummies(train_df, columns=['Type'], drop_first=True)
val_df = pd.get_dummies(val_df, columns=['Type'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Type'], drop_first=True)

# Alinear columnas entre conjuntos
train_df, val_df = train_df.align(val_df, join='left', axis=1, fill_value=0)
train_df, test_df = train_df.align(test_df, join='left', axis=1, fill_value=0)

# Asegurar que val_df y test_df queden exactamente con las mismas columnas que train_df
val_df = val_df.reindex(columns=train_df.columns, fill_value=0)
test_df = test_df.reindex(columns=train_df.columns, fill_value=0)
```

```python
# Detectar automáticamente columnas dummy de Type
type_dummies = [col for col in train_df.columns if col.startswith('Type_')]

features = [
    'Store', 'Dept', 'Size',
    'Temperature', 'Fuel_Price', 'CPI', 'Unemployment',
    'IsHoliday',
    'Year', 'Month', 'Week'
]

# Agregar dummies de Type
features += type_dummies

# Agregar variables binarias de markdown
features += [f'{col}_bin' for col in markdown_cols]

print("Variables seleccionadas para el modelo:")
print(features)
```

```python
X_train = train_df[features]
y_train = train_df['Weekly_Sales']

X_val = val_df[features]
y_val = val_df['Weekly_Sales']

X_test = test_df[features]
y_test = test_df['Weekly_Sales']

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

### 5.3 Modelo base: Regresión Lineal

Se utiliza un modelo de regresión lineal como referencia inicial, debido a su simplicidad e interpretabilidad.

Este modelo permite establecer una línea base frente a la cual comparar modelos más complejos y flexibles.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

model_lr = LinearRegression()
model_lr.fit(X_train, y_train)

y_pred_val_lr = model_lr.predict(X_val)

mae_lr = mean_absolute_error(y_val, y_pred_val_lr)
rmse_lr = np.sqrt(mean_squared_error(y_val, y_pred_val_lr))

print("Linear Regression - Validation MAE:", mae_lr)
print("Linear Regression - Validation RMSE:", rmse_lr)
```

### 5.4 Modelo alternativo: Random Forest

Se utiliza un modelo Random Forest Regressor, el cual permite capturar relaciones no lineales e interacciones entre variables, siendo adecuado para datasets heterogéneos como el presente caso.

Este modelo es menos sensible a supuestos de linealidad y puede adaptarse mejor a fenómenos complejos presentes en las ventas semanales.

```python
from sklearn.ensemble import RandomForestRegressor

model_rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model_rf.fit(X_train, y_train)

y_pred_val_rf = model_rf.predict(X_val)

mae_rf = mean_absolute_error(y_val, y_pred_val_rf)
rmse_rf = np.sqrt(mean_squared_error(y_val, y_pred_val_rf))

print("Random Forest - Validation MAE:", mae_rf)
print("Random Forest - Validation RMSE:", rmse_rf)
```

### 5.5 Comparación de resultados


```python
print("| Modelo           |   MAE   |  RMSE  |")
print("|------------------|---------|--------|")
print(f"| Regresión Lineal | {mae_lr:7,.0f} | {rmse_lr:6,.0f} |")
print(f"| Random Forest    | {mae_rf:7,.0f} | {rmse_rf:6,.0f} |")
```

#### Observaciones

Los resultados evidencian una diferencia significativa en el desempeño de los modelos evaluados.

El modelo de Regresión Lineal presenta errores considerablemente mayores, lo que indica que no logra capturar adecuadamente la complejidad del problema. Esto sugiere que las relaciones entre las variables explicativas y las ventas no son puramente lineales.

Por otro lado, el modelo Random Forest muestra una mejora sustancial en ambas métricas (MAE y RMSE), reduciendo el error en más de un 80% respecto al modelo lineal. Este resultado confirma que el problema presenta relaciones no lineales e interacciones entre variables, las cuales son capturadas de mejor forma por modelos basados en árboles.

Adicionalmente, este resultado valida las decisiones tomadas en etapas anteriores, particularmente:

- La transformación de variables MarkDown a formato binario
- La inclusión de variables temporales
- El uso de variables categóricas codificadas

En conjunto, estos elementos permiten al modelo capturar patrones relevantes del comportamiento de las ventas.

Por lo tanto, se selecciona el modelo Random Forest como modelo base para la etapa de evaluación final.
